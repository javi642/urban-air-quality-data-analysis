## IMPORTING LIBRARY FUNCTIONS


```python
# Loads required libraries for data analysis and visualization.
# Prepares the environment for creating insightful charts and plots.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as pxq
```

## LOAD DATASET

```python
# Loads the air pollution dataset from an Excel file.
# Stores the data in a DataFrame for further analysis and visualization.

df = pd.read_excel(r"C:\Users\suvin\Downloads\UrbanAirPollutionDataset.xlsx")
```

## 1.SCATTER PLOT- PM2.5 VS AQI

```python
# Randomly samples 10 records to keep the scatter plot clean and readable.
# Visualizes the relationship between PM2.5 levels and the AQI target using a scatter plot.

sample_df = df.sample(10, random_state=42)
plt.figure(figsize=(10,6))

plt.scatter(
    sample_df["PM2.5"],
    sample_df["AQI_Target"],
    s=100,        # large point size
    color="brown", # same style
)
plt.title("PM2.5 vs AQI (Sampled Scatter Plot)", fontsize=16)
plt.xlabel("PM2.5", fontsize=12)
plt.ylabel("AQI", fontsize=12)
plt.grid(False)  # no grid (like your image)
plt.show()
```

## 2. BAR CHART — Average Pollutant Levels

```python
# Computes the mean levels of major air pollutants.
# Displays their average concentrations using a bar chart for quick comparison.

pollutants = ["PM2.5", "PM10", "NO₂", "SO₂", "CO", "O₃"]
avg_values = df[pollutants].mean()
plt.figure(figsize=(8,5))
sns.barplot(x=pollutants, y=avg_values)
plt.title("Average Levels of Pollutants")
plt.xlabel("Pollutants")
plt.ylabel("Average Level")
plt.show()
```

## 3. LINE CHART — AQI Over Time

```python
# Converts the DateTime column to proper datetime format for time-series analysis.
# Samples the data at regular intervals and plots a clean AQI trend over time.

df["DateTime"] = pd.to_datetime(df["DateTime"])
df_sampled = df.iloc[::200]
plt.figure(figsize=(12,5))
plt.plot(df_sampled["DateTime"], df_sampled["AQI_Target"], color="blue", linewidth=2)
plt.title("AQI Trend Over Time (Clean Sampled Plot)", fontsize=16)
plt.xlabel("Date & Time")
plt.ylabel("AQI")
plt.grid(True)
plt.tight_layout()
plt.show()

```

## 4. HEATMAP — Correlation Between Variables

```python
# Analyzes correlations among air quality variables.
# Displays the strength and direction of relationships using a correlation heatmap.

plt.figure(figsize=(10,7))
sns.heatmap(df.corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap of Air Quality Features")
plt.show()
```

## 5. PIE CHART — Pollutant Contribution

```python
# Calculates the average levels of major air pollutants.
# Visualizes their percentage contribution using a pie chart.

pollutants = ["PM2.5", "PM10", "NO₂", "SO₂", "CO", "O₃"]
```

```python
avg_values = df[pollutants].mean()
```

```python
plt.figure(figsize=(8,8))

plt.pie(
    avg_values,
    labels=pollutants,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Pollutant Contribution (%)", fontsize=16)
plt.show()
```

## HISTOGRAM-Frequency Distribution of PM2.5 Pollutant Values

```python
# Examines how PM2.5 values are distributed across observations.
# Uses a histogram to show their frequency and spread.

plt.figure(figsize=(8,5))
sns.histplot(df["PM2.5"], bins=30, color="red")
plt.title("Histogram of PM2.5 Levels", fontsize=16)
plt.xlabel("PM2.5 Value")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

## HISTOGRAM-Distribution of All Pollutant Concentration Levels

```python
# Examines the distribution of key air pollutants.
# Visualizes their frequency using separate histograms arranged in subplots.

pollutants = ["PM2.5", "PM10", "NO₂", "SO₂", "CO", "O₃"]
plt.figure(figsize=(12,10))
for i, col in enumerate(pollutants, 1):
    plt.subplot(3, 2, i)
    sns.histplot(df[col], bins=30, kde=True, color="blue")
    plt.title(f"Histogram of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

## Inclusion of Additional Pollutant Metrics: NH₃ and Benzene

```python
# ===============================
# ADD TWO NEW POLLUTANTS
# NH3 and Benzene
# ===============================

# NH3 (Ammonia): realistic range 10–60 µg/m³
df["NH3"] = np.random.uniform(10, 60, size=len(df))

# Benzene: realistic range 0.5–9.0 µg/m³
df["Benzene"] = np.random.uniform(0.5, 9.0, size=len(df))

# ===============================
# CONFIRM VALUES HAVE BEEN ADDED
# ===============================
print("New pollutant columns added successfully!\n")

# Show first 10 rows
print("Preview of new columns:\n")
print(df[["NH3", "Benzene"]].head(10))

# Show shape to confirm 50,000 rows updated
print("\nDataset shape (rows, columns):", df.shape)
```
